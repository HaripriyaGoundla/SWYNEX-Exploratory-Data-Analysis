# SWYNEX – Exploratory Data Analysis

**Task 2 | Dataset:** cleaned_customer_data.csv

This notebook explores customer records using pandas and matplotlib. The dataset is small (10 rows), so findings are descriptive and should not be generalized without more data.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned Task 1 dataset
df = pd.read_csv('cleaned_customer_data.csv')
df['signup date'] = pd.to_datetime(df['signup date'], errors='coerce')
df.head()
```

## 1. Dataset overview

```python
print('Rows, columns:', df.shape)
df.info()
display(df.isna().sum().to_frame('Missing values'))
```

## 2. Descriptive statistics

```python
display(df.describe(include='all').T)
print('Median amount:', df['amount'].median())
print('Total amount:', df['amount'].sum())
```

## 3. Customer distribution by country

```python
country_counts = df['country'].fillna('Missing').value_counts()
country_counts.plot(kind='bar', figsize=(9,5), title='Customer Count by Country')
plt.xlabel('Country'); plt.ylabel('Customers'); plt.xticks(rotation=35, ha='right'); plt.tight_layout(); plt.show()
country_counts
```

## 4. Active vs inactive customers

```python
activity = df['is_active'].value_counts().reindex([True, False], fill_value=0)
activity.index = ['Active', 'Inactive']
activity.plot(kind='pie', autopct='%1.0f%%', figsize=(6,5), title='Customer Activity Status')
plt.ylabel(''); plt.show()
print('Active percentage:', df['is_active'].mean()*100)
```

## 5. Amount distribution and anomaly checks

```python
df['amount'].dropna().plot(kind='hist', bins=6, edgecolor='black', figsize=(7,5), title='Distribution of Amounts')
plt.xlabel('Amount'); plt.ylabel('Frequency'); plt.show()
print(df['amount'].describe())
print('Negative amount records:')
display(df[df['amount'] < 0])
```

## 6. Signup trend

```python
valid_dates = df.dropna(subset=['signup date']).copy()
monthly = valid_dates.groupby(valid_dates['signup date'].dt.to_period('M')).size()
monthly.index = monthly.index.astype(str)
monthly.plot(kind='line', marker='o', figsize=(8,5), title='Customer Signups by Month')
plt.xlabel('Month'); plt.ylabel('Signups'); plt.xticks(rotation=35, ha='right'); plt.tight_layout(); plt.show()
monthly
```

## 7. Key findings

1. The dataset contains 10 customer records and 8 columns.
2. 7/10 customers are active (70%); 3/10 are inactive (30%).
3. United States is the most represented named country (2 customers); one country is missing.
4. 7 of 9 valid signup dates are in January 2026 (77.8%); May and October have one each.
5. Amount mean is 1,657.83 and median is 1,200.00; high values influence the mean.
6. One negative amount (-30.00) is flagged; validate whether it is a refund, adjustment, or error.
7. Full Name, signup date, amount, and country each have one missing value.
8. The maximum amount is 5,000.00 for the United Kingdom record; investigate context before classifying it as an anomaly.

**Limitation:** Only 10 rows are available, so these are descriptive observations, not statistically generalizable conclusions.
